```python
import pandas as pd
import numpy as np

data = {
    "age": [25, 32, np.nan, 45, 29],
    "salary": [50000, 65000, 70000, np.nan, 52000],
    "city": ["Hyderabad", "Bangalore", "Hyderabad", "Chennai", None],
    "purchased": ["Yes", "No", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

df
```

```python
df.isnull().sum()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df["salary"] = df["salary"].fillna(df["salary"].median())
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
```

```python
df.isnull().sum()
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=["city"],
    drop_first=False
)

encoded_df
```

```python
encoded_df["purchased"] = encoded_df["purchased"].map({
    "No": 0,
    "Yes": 1
})
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

encoded_df[["age", "salary"]] = scaler.fit_transform(
    encoded_df[["age", "salary"]]
)

encoded_df
```
